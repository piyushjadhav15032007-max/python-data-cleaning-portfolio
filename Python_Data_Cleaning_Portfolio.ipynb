```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Customer": ["Asha", "Rahul", "Neha", "Asha", "Vijay", "Meera"],
    "City": ["Pune", "Mumbai", None, "Pune", "Nashik", "Nagpur"],
    "Sales": [1200, 1500, 900, 1200, None, 1600]
}

df = pd.DataFrame(data)

print("Original data:")
display(df)

cleaned = df.drop_duplicates().copy()
cleaned["City"] = cleaned["City"].fillna("Unknown")

print("Cleaned data:")
display(cleaned)

print("Duplicate rows removed:", len(df) - len(cleaned))
print("Missing values after cleaning:")
display(cleaned.isna().sum())

print("Total available sales:", cleaned["Sales"].sum())
print("Average available sales:", cleaned["Sales"].mean())

cleaned.groupby("City")["Sales"].sum().plot(kind="bar")
plt.title("Available Sales by City")
plt.ylabel("Sales")
plt.show()
```
